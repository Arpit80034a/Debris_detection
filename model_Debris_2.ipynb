```python
import zipfile

with zipfile.ZipFile("Debris_detection.yolov9.zip", 'r') as zip_ref:
    zip_ref.extractall("/content/")
```

```python
import yaml
import os

# 1. Define the correct paths
dataset_path = "/content/Debris_detection.yolov9"
yaml_path = "/content/data.yaml"

# 2. Check what folders actually exist to avoid the FileNotFoundError
# Common Roboflow names are 'train', 'valid', and 'test'
train_path = os.path.join(dataset_path, "train/images")
val_path = os.path.join(dataset_path, "valid/images")
test_path = os.path.join(dataset_path, "test/images")

# Fallback: if 'valid' doesn't exist, check for 'val'
if not os.path.exists(val_path):
    val_path = os.path.join(dataset_path, "val/images")

# 3. Update the YAML file
data = {
    'train': train_path,
    'val': val_path,
    'test': test_path,
    'nc': 1, # Update this to your actual number of classes
    'names': ['debris'] # Update this to your actual class names
}

with open(yaml_path, 'w') as f:
    yaml.dump(data, f)

print(f"YAML updated. Validation path set to: {val_path}")
```

```python
from ultralytics import YOLO

# Load pretrained model
model = YOLO("yolov8n.pt")  # n = nano (fast), can use s/m/l/x

# Train the model
model.train(
    data="data.yaml",
    epochs=50,
    imgsz=640,
    batch=16,
    name="debris_model"
)
```
