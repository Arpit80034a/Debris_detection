```python
import re
import pandas as pd

epochs = []
box_loss = []
cls_loss = []
dfl_loss = []
precision = []
recall = []
map50 = []
map5095 = []

lines = log_text.split("\n")

for i, line in enumerate(lines):
    epoch_match = re.search(r"(\d+)/\d+", line)

    if epoch_match:
        epoch = int(epoch_match.group(1))

        try:
            parts = re.findall(r"\d+\.\d+", line)

            if len(parts) >= 3:
                epochs.append(epoch)
                box_loss.append(float(parts[0]))
                cls_loss.append(float(parts[1]))
                dfl_loss.append(float(parts[2]))

                # Next lines contain metrics
                for j in range(i, min(i+10, len(lines))):
                    if "all" in lines[j]:
                        vals = re.findall(r"\d+\.\d+", lines[j])
                        if len(vals) >= 4:
                            precision.append(float(vals[0]))
                            recall.append(float(vals[1]))
                            map50.append(float(vals[2]))
                            map5095.append(float(vals[3]))
                        break
        except:
            continue

df = pd.DataFrame({
    "epoch": epochs,
    "box_loss": box_loss,
    "cls_loss": cls_loss,
    "dfl_loss": dfl_loss,
    "precision": precision,
    "recall": recall,
    "mAP50": map50,
    "mAP50-95": map5095
})

df.head()
```

```python
import matplotlib.pyplot as plt

# ---------- LOSS PLOTS ----------
plt.figure()
plt.plot(df["epoch"], df["box_loss"], label="Box Loss")
plt.plot(df["epoch"], df["cls_loss"], label="Class Loss")
plt.plot(df["epoch"], df["dfl_loss"], label="DFL Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Losses vs Epoch")
plt.legend()
plt.grid()
plt.show()



plt.savefig("graph.png")


# ---------- PRECISION & RECALL ----------
plt.figure()
plt.plot(df["epoch"], df["precision"], label="Precision (P)")
plt.plot(df["epoch"], df["recall"], label="Recall (R)")

plt.xlabel("Epoch")
plt.ylabel("Score")
plt.title("Precision & Recall vs Epoch")
plt.legend()
plt.grid()
plt.show()


plt.savefig("graph.png")


# ---------- MAP SCORES ----------
plt.figure()
plt.plot(df["epoch"], df["mAP50"], label="mAP@50")
plt.plot(df["epoch"], df["mAP50-95"], label="mAP@50-95")

plt.xlabel("Epoch")
plt.ylabel("mAP")
plt.title("mAP Scores vs Epoch")
plt.legend()
plt.grid()
plt.show()



plt.savefig("graph.png")
```

```python
df["total_loss"] = df["box_loss"] + df["cls_loss"] + df["dfl_loss"]

plt.figure()
plt.plot(df["epoch"], df["total_loss"], label="Total Loss")

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Total Loss vs Epoch")
plt.legend()
plt.grid()
plt.show()


plt.savefig("graph.png")
```

```python
df["f1_score"] = 2 * (df["precision"] * df["recall"]) / (df["precision"] + df["recall"])

plt.figure()
plt.plot(df["epoch"], df["f1_score"], label="F1 Score")

plt.xlabel("Epoch")
plt.ylabel("F1 Score")
plt.title("F1 Score vs Epoch")
plt.legend()
plt.grid()
plt.show()

plt.savefig("graph.png")
```
